```python
from pyspark.sql import functions as F
SOURCE = "s3a://data5035-spring26/drone_data.json"

source_df = spark.read.json(SOURCE)
display(source_df.limit(10)); 

```

### The Calibration tables 
Contains a single row for each CESIUM_137_DETECTOR_UNIT_NUMBER with CESIUM_137_DETECTOR_CALIBRATION_PRECISION and CESIUM_137_DETECTOR_CALIBRATION_TIMESTAMP
- UNIT_NUMBER would be the primary identifier. 
- A key can be make
*note I would make a table of these for each detector unit. 

### The Levels table
Contains a single row for each CESIUM_137_DETECTOR_UNIT_NUMBER with 
CESIUM_137_LEVEL and COLLECTION TIME
- UNIT_NUMBER WILL BE THE KEY AGAIN

### The Units table
- Contains a simgle row for each distinct unit the CESIUM_137_DETECTOR_UNIT_NUMBER appended with rows for each distinct unit in both the GAMMA_DETECTOR_UNIT_NUMBER & THORIUM_232_DETECTOR_UNIT_NUMBER


```python
source_df.createOrReplaceTempView("gamma_detector_levels")
spark.sql("""
          SELECT 
            DISTINCT(GAMMA_DETECTOR_UNIT_NUMBER),
            AVG(GAMMA_DETECTOR_CALIBRATION_PRECISION)
          FROM gamma_detector_levels
          GROUP BY GAMMA_DETECTOR_UNIT_NUMBER"""
).display()
```
